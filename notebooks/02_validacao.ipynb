# 02 - Validação

Gera o relatório de validação (módulo 2) para cada planilha e a checagem de divergência entre as colunas H-max de mesmo nome nas duas planilhas.

```python
from IPython.display import Markdown, display

from harmonic_signatures.io.cache import read_workbook_cached
from harmonic_signatures.config import DEFAULT_WORKBOOK, INTERIM_DIR, DEFAULT_JOIN_TOLERANCE_SECONDS
from harmonic_signatures.validation.checks import validate_sheet, compare_h_max_across_sheets

result = read_workbook_cached(DEFAULT_WORKBOOK, INTERIM_DIR)
```

## Relatório — HARMÔNICAS

```python
harm_report = validate_sheet(result.harmonicas)
display(Markdown(harm_report.to_markdown()))
```

## Relatório — INTERHARMÔNICAS

```python
inter_report = validate_sheet(result.interharmonicas)
display(Markdown(inter_report.to_markdown()))
```

## Divergência das colunas H-max de mesmo nome

As colunas `H00`..`H50 max` aparecem com o mesmo nome nas duas planilhas, mas **não são duplicatas**: a comparação abaixo confirma empiricamente (baixa correlação, sem razão de escala fixa) que são fontes de dado distintas. Por isso ambas são mantidas como features separadas no dataset combinado (módulo de organização), marcadas por `source_sheet`.

```python
cross_report = compare_h_max_across_sheets(
    result.harmonicas, result.interharmonicas,
    tolerance_seconds=DEFAULT_JOIN_TOLERANCE_SECONDS,
)
display(Markdown(cross_report.to_markdown()))
```
