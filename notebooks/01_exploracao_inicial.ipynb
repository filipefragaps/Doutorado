# 01 - Exploração inicial

Importa o banco de dados consolidado (planilhas HARMÔNICAS e INTERHARMÔNICAS) usando o pacote `harmonic_signatures` e inspeciona a estrutura resultante.

```python
from harmonic_signatures.io.cache import read_workbook_cached
from harmonic_signatures.config import DEFAULT_WORKBOOK, INTERIM_DIR

result = read_workbook_cached(DEFAULT_WORKBOOK, INTERIM_DIR)
print('Origem:', result.source_path)
print('Importado em:', result.imported_at)
print('Colunas não reconhecidas:', result.unknown_columns)
```

## Planilha HARMÔNICAS

```python
harm = result.harmonicas
print('shape:', harm.dataframe.shape)
print('label_column:', harm.label_column, '| timestamp_column:', harm.timestamp_column, '| status_column:', harm.status_column)
print('n medições:', len(harm.measurement_columns()))
harm.dataframe[harm.label_column].value_counts()
```

## Planilha INTERHARMÔNICAS

```python
inter = result.interharmonicas
print('shape:', inter.dataframe.shape)
print('label_column:', inter.label_column, '| timestamp_column:', inter.timestamp_column)
print('n medições:', len(inter.measurement_columns()))
inter.dataframe[inter.label_column].value_counts()
```

## Famílias de colunas reconhecidas em cada planilha

```python
import pandas as pd

for sheet in (harm, inter):
    families = pd.Series([s.family for s in sheet.columns.values() if s.kind == 'measurement'])
    print(sheet.name)
    print(families.value_counts())
    print()
```
